```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
df.info()
```

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})

missing
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:", duplicate_count)
print("Percentage of duplicate records:",
      round(duplicate_count / len(df) * 100, 2), "%")
```

```python
duplicates = df[df.duplicated(keep=False)].sort_values(
    by=["InvoiceNo", "StockCode", "InvoiceDate"]
)

duplicates.head(20)
```

```python
print("Unique countries:", df["Country"].nunique())
print("\nCountry values:")
print(sorted(df["Country"].dropna().unique()))
```

```python
print("Descriptions with leading/trailing spaces:",
      df["Description"].fillna("").ne(
          df["Description"].fillna("").str.strip()
      ).sum())

print("Blank descriptions:",
      df["Description"].fillna("").str.strip().eq("").sum())
```

```python
print("Negative quantities:", (df["Quantity"] < 0).sum())
print("Zero quantities:", (df["Quantity"] == 0).sum())
print("Negative prices:", (df["UnitPrice"] < 0).sum())
print("Zero prices:", (df["UnitPrice"] == 0).sum())
```

```python
negative_qty = df[df["Quantity"] < 0]

print("Negative-quantity records:", len(negative_qty))
print("\nInvoices beginning with C:",
      negative_qty["InvoiceNo"].astype(str).str.startswith("C").sum())

print("\nSample negative transactions:")
negative_qty.head(10)
```

```python
negative_prices = df[df["UnitPrice"] < 0]

print("Negative-price records:", len(negative_prices))
negative_prices
```

```python
zero_price = df[df["UnitPrice"] == 0]

print("Zero-price records:", len(zero_price))
print("\nDescriptions of zero-price records:")
print(zero_price["Description"].value_counts().head(20))

print("\nSample zero-price records:")
zero_price.head(10)
```

```python
print("Invoice number patterns:")

invoice_str = df["InvoiceNo"].astype(str)

print("Invoices starting with C:",
      invoice_str.str.startswith("C").sum())

print("Invoices starting with A:",
      invoice_str.str.startswith("A").sum())

print("Numeric invoices:",
      invoice_str.str.match(r"^\d+$").sum())
```

```python
adjustments = df[df["InvoiceNo"].astype(str).str.startswith("A")]

adjustments
```

```python
cleaned_df = df.copy()

print("Original rows:", len(df))
print("Cleaning copy created:", len(cleaned_df))
```

```python
before = len(cleaned_df)

cleaned_df = cleaned_df.drop_duplicates().copy()

after = len(cleaned_df)

print("Rows before removing duplicates:", before)
print("Duplicate rows removed:", before - after)
print("Rows after removing duplicates:", after)
```

```python
cleaned_df["Description"] = (
    cleaned_df["Description"]
    .str.strip()
    .str.upper()
)

print("Description whitespace cleaned successfully.")
print("Missing descriptions:", cleaned_df["Description"].isna().sum())
```

```python
cleaned_df.info()
```

```python
before = len(cleaned_df)

cleaned_df = cleaned_df[
    ~cleaned_df["InvoiceNo"].astype(str).str.startswith("A")
].copy()

after = len(cleaned_df)

print("Adjustment records removed:", before - after)
print("Rows after removing adjustments:", after)
```

```python
cleaned_df["IsCancellation"] = (
    cleaned_df["InvoiceNo"].astype(str).str.startswith("C")
)

print("Cancellation records:", cleaned_df["IsCancellation"].sum())
print("Regular transaction records:",
      (~cleaned_df["IsCancellation"]).sum())
```

```python
print("Missing descriptions:", cleaned_df["Description"].isna().sum())
print("Percentage missing:",
      round(cleaned_df["Description"].isna().mean() * 100, 2), "%")
```

```python
print("Missing CustomerIDs:", cleaned_df["CustomerID"].isna().sum())
print("Percentage missing:",
      round(cleaned_df["CustomerID"].isna().mean() * 100, 2), "%")
```

```python
missing_customer = cleaned_df[cleaned_df["CustomerID"].isna()]

print("Records with missing CustomerID:", len(missing_customer))
print("Cancellation records among them:",
      missing_customer["IsCancellation"].sum())
print("Percentage of missing CustomerID records that are cancellations:",
      round(
          missing_customer["IsCancellation"].mean() * 100, 2
      ), "%")
```

```python
cleaned_df["CustomerID"] = cleaned_df["CustomerID"].astype("Int64")

print("CustomerID data type:", cleaned_df["CustomerID"].dtype)
```

```python
print("Negative quantities:", (cleaned_df["Quantity"] < 0).sum())
print("Zero quantities:", (cleaned_df["Quantity"] == 0).sum())
print("Negative prices:", (cleaned_df["UnitPrice"] < 0).sum())
print("Zero prices:", (cleaned_df["UnitPrice"] == 0).sum())
```

```python
zero_price = cleaned_df[cleaned_df["UnitPrice"] == 0]

print("Zero-price records:", len(zero_price))
print("\nTop descriptions:")
print(zero_price["Description"].value_counts(dropna=False).head(15))
```

```python
quality_summary = pd.DataFrame({
    "Issue": [
        "Missing Description",
        "Missing CustomerID",
        "Duplicate Records",
        "Negative Quantity (cancellations/returns)",
        "Zero Quantity",
        "Negative UnitPrice",
        "Zero UnitPrice"
    ],
    "Count": [
        cleaned_df["Description"].isna().sum(),
        cleaned_df["CustomerID"].isna().sum(),
        cleaned_df.duplicated().sum(),
        (cleaned_df["Quantity"] < 0).sum(),
        (cleaned_df["Quantity"] == 0).sum(),
        (cleaned_df["UnitPrice"] < 0).sum(),
        (cleaned_df["UnitPrice"] == 0).sum()
    ]
})

quality_summary
```

```python
print("Final rows:", len(cleaned_df))
print("Final columns:", len(cleaned_df.columns))

print("\nColumns:")
print(cleaned_df.columns.tolist())

print("\nData types:")
print(cleaned_df.dtypes)
```

```python
cleaned_df = cleaned_df.reset_index(drop=True)

print("Index reset successfully!")
print("First 5 index values:", cleaned_df.index[:5].tolist())
print("Last index value:", cleaned_df.index[-1])
```

```python
print("Remaining duplicate records:", cleaned_df.duplicated().sum())
```

```python
final_summary = pd.DataFrame({
    "Metric": [
        "Original rows",
        "Final rows",
        "Rows removed",
        "Duplicate rows removed",
        "Adjustment records removed",
        "Missing Description",
        "Missing CustomerID",
        "Cancellation/return records",
        "Zero-price records"
    ],
    "Value": [
        len(df),
        len(cleaned_df),
        len(df) - len(cleaned_df),
        5268,
        3,
        cleaned_df["Description"].isna().sum(),
        cleaned_df["CustomerID"].isna().sum(),
        cleaned_df["IsCancellation"].sum(),
        (cleaned_df["UnitPrice"] == 0).sum()
    ]
})

final_summary
```

```python
cleaned_df.to_csv("SWYNEX_Online_Retail_Cleaned.csv", index=False)

print("Cleaned dataset exported successfully!")
print("Rows exported:", len(cleaned_df))
print("File: SWYNEX_Online_Retail_Cleaned.csv")
```
